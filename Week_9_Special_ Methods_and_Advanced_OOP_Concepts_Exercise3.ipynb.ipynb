```python

# Book class
class Book:

    def __init__(self, title, author):
        self.title = title
        self.author = author

    # Return book details
    def get_details(self):
        return f"Title: {self.title}\nAuthor: {self.author}"


# Library class
class Library:

    def __init__(self, name):
        self.name = name

        # Create two Book objects
        self.books = [
            Book("Introduction to Pythont", "John Smith"),
            Book("Data Structurese", "Mary Jones")
        ]

    # Display all books
    def display_books(self):
        print("LIBRARY NAME:", self.name)
        print("=" * 40)

        for book in self.books:
            print(book.get_details())
            print("-" * 40)


# Create a Library object
my_library = Library("Kabarak University Library")

# Display all books
my_library.display_books()
```
