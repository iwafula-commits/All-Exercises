```python

# CPU class
class CPU:

    def __init__(self, brand, core_count):
        self.brand = brand
        self.core_count = core_count

    # Method to return CPU specifications
    def get_specs(self):
        return f"CPU Brand: {self.brand}\nCPU Cores: {self.core_count}"


# Laptop class using composition
class Laptop:

    def __init__(self, model, ram, cpu):
        self.model = model
        self.ram = ram
        self.cpu = cpu  # Laptop contains a CPU object

    # Display complete laptop information
    def display_info(self):
        print("LAPTOP SPECIFICATIONS")
        print("-------------------------")
        print("Laptop Model:", self.model)
        print("RAM:", self.ram)
        print(self.cpu.get_specs())


# Create CPU objects
cpu1 = CPU("Intel Core i7", 8)
cpu2 = CPU("AMD Ryzen 7", 8)

# Create Laptop objects using CPU objects
laptop1 = Laptop("Dell Latitude 5420", "16GB", cpu1)
laptop2 = Laptop("HP EliteBook 840", "8GB", cpu2)

# Test the objects
laptop1.display_info()

print("\n")

laptop2.display_info()
```
