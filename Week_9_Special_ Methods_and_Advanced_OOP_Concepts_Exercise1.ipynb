```python

class ShoppingCart:

    def __init__(self, items=None):
        self.items = items if items is not None else []

    # Returns the total number of items
    def __len__(self):
        return len(self.items)

    # Returns a friendly summary
    def __str__(self):
        if not self.items:
            return "Your shopping cart is empty."

        summary = "Shopping Cart Items:\n"
        for item in self.items:
            summary += f"- {item}\n"

        summary += f"Total items: {len(self.items)}"
        return summary

    # Overload + operator to merge two carts
    def __add__(self, other):
        if not isinstance(other, ShoppingCart):
            return NotImplemented

        merged_items = self.items + other.items
        return ShoppingCart(merged_items)


# Create the first shopping cart
cart1 = ShoppingCart(["Laptop", "Mouse", "Keyboard"])

# Create the second shopping cart
cart2 = ShoppingCart(["Phone", "Headphones", "Charger"])

# Display the first cart
print("FIRST CART")
print(cart1)

# Display the second cart
print("\nSECOND CART")
print(cart2)

# Display number of items
print("\nItems in first cart:", len(cart1))
print("Items in second cart:", len(cart2))

# Merge the two shopping carts
cart3 = cart1 + cart2

# Display merged cart
print("\nMERGED SHOPPING CART")
print(cart3)

# Display total number of merged items
print("\nTotal items in merged cart:", len(cart3))
```
